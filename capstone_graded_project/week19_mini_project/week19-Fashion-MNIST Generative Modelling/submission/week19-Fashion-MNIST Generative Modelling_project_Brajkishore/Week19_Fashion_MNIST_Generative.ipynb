## 2. Approach and Model Design

We implemented a compact DCGAN-like architecture suitable for Fashion-MNIST: a `Generator` that maps a 100-dim latent vector to a 28x28 image via a linear layer and transpose-convolutions, and a `Discriminator` that uses convolutional downsampling to a single logit. The design choices prioritize simplicity and fast iteration on CPU.

### Design Decisions (systematic explanation)

- **Model simplicity**: small architectures reduce training time and overfitting on small images.
- **Latent dim = 100**: standard choice balancing capacity and stability.
- **BatchNorm and LeakyReLU**: used for stable GAN training.
- **Tanh output + normalization**: images normalized to [-1,1] for generator; denormalized for visualization.
- **Loss**: `BCEWithLogitsLoss` for discriminator and generator objectives. This is a conventional baseline for DCGANs.
- **Optimizer**: Adam with betas (0.5, 0.999) as commonly used in GAN literature.
- **Checkpointing**: save generator and discriminator state after each epoch to `solution/output` for reproducibility.
