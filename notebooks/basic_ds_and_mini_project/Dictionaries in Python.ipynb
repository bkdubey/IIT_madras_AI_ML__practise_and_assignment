## 1. Additional Dictionary Methods

Beyond basic access and modification, dictionaries have many powerful methods.

### get() - Safe Access with Default

**Syntax:** `dict.get(key, default=None)`

Returns value for key, or default if key doesn't exist (no error!)

````python
# get() is safer than direct access
student = {'name': 'Alice', 'age': 20, 'grade': 'A'}

# Regular access
print("Name:", student['name'])

# Using get() - returns None if not found
print("Email:", student.get('email'))  # Returns None, no error!

# Using get() with default value
print("Email with default:", student.get('email', 'not provided'))
print("Age with default:", student.get('age', 0))
````

### setdefault() - Get or Set

**Syntax:** `dict.setdefault(key, default=None)`

Returns value if key exists, otherwise sets and returns default

````python
# setdefault() gets or creates key
student = {'name': 'Bob', 'age': 21}

print("Before:", student)

# Get existing key
name = student.setdefault('name', 'Unknown')
print("Name:", name)

# Create new key with default
grade = student.setdefault('grade', 'Not assigned')
print("Grade:", grade)
print("After:", student)
````

### pop() - Remove and Return

**Syntax:** `dict.pop(key, default=None)`

````python
# pop() removes key and returns value
student = {'name': 'Charlie', 'age': 22, 'grade': 'B', 'email': 'charlie@email.com'}

print("Before:", student)

# Remove and get value
email = student.pop('email')
print("Removed email:", email)
print("After:", student)

# Pop with default (if key doesn't exist)
phone = student.pop('phone', 'No phone')
print("Phone:", phone)
````

### popitem() - Remove Last Item

**Syntax:** `dict.popitem()`

Removes and returns last (key, value) pair

````python
# popitem() removes last inserted item
student = {'name': 'David', 'age': 23, 'grade': 'A'}

print("Before:", student)

last_item = student.popitem()
print("Removed:", last_item)
print("After:", student)
````

### clear() - Remove All Items

**Syntax:** `dict.clear()`

````python
# clear() empties the dictionary
student = {'name': 'Eve', 'age': 24, 'grade': 'B'}

print("Before:", student)
student.clear()
print("After:", student)
````

### copy() - Create a Copy

**Syntax:** `new_dict = dict.copy()`

````python
# copy() creates an independent copy
original = {'name': 'Frank', 'age': 25}
copy_dict = original.copy()

print("Original:", original)
print("Copy:", copy_dict)

# Modifying copy doesn't affect original
copy_dict['grade'] = 'A'
print("\nAfter modifying copy:")
print("Original:", original)
print("Copy:", copy_dict)
````

## 2. Viewing Dictionary Contents

Methods to view keys, values, and items.

### keys() - Get All Keys

````python
student = {'name': 'Alice', 'age': 20, 'grade': 'A', 'major': 'CS'}

# Get all keys
keys = student.keys()
print("Keys:", keys)
print("Type:", type(keys))

# Convert to list
keys_list = list(student.keys())
print("Keys as list:", keys_list)
````

### values() - Get All Values

````python
student = {'name': 'Bob', 'age': 21, 'grade': 'B', 'major': 'Math'}

# Get all values
values = student.values()
print("Values:", values)

# Convert to list
values_list = list(student.values())
print("Values as list:", values_list)
````

### items() - Get Key-Value Pairs

````python
student = {'name': 'Charlie', 'age': 22, 'grade': 'A'}

# Get all key-value pairs
items = student.items()
print("Items:", items)

# Convert to list of tuples
items_list = list(student.items())
print("Items as list:", items_list)
print("First item:", items_list[0])
````

## 4. Updating Dictionaries

Multiple ways to update dictionary contents.

### update() - Merge Dictionaries

**Syntax:** `dict1.update(dict2)`

````python
# update() merges another dictionary
student = {'name': 'Alice', 'age': 20}
additional_info = {'grade': 'A', 'major': 'CS'}

print("Before update:", student)
student.update(additional_info)
print("After update:", student)
````

````python
# update() can overwrite existing keys
student = {'name': 'Bob', 'age': 21, 'grade': 'B'}
updates = {'age': 22, 'grade': 'A'}  # Updating existing keys

print("Before:", student)
student.update(updates)
print("After:", student)
````

### Merging with | Operator (Python 3.9+)

**Syntax:** `dict3 = dict1 | dict2`

````python
# Merge dictionaries with | operator
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

merged = dict1 | dict2
print("Merged:", merged)
print("Original dict1:", dict1)  # Unchanged
````

````python
# Right side wins on conflicts
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 20, 'c': 3}  # 'b' exists in both

merged = dict1 | dict2
print("Merged:", merged)  # dict2's 'b' value wins
````

## 5. Nested Dictionaries

Dictionaries can contain other dictionaries, creating hierarchical structures.

### Creating Nested Dictionaries

````python
# Store student information with nested dictionaries
students = {
    'student1': {
        'name': 'Alice',
        'age': 20,
        'grades': {'math': 90, 'science': 85}
    },
    'student2': {
        'name': 'Bob',
        'age': 21,
        'grades': {'math': 85, 'science': 90}
    }
}

print("All students:", students)
````

### Accessing Nested Values

Use multiple brackets: `dict[key1][key2]`

````python
students = {
    'student1': {
        'name': 'Alice',
        'age': 20,
        'grades': {'math': 90, 'science': 85}
    },
    'student2': {
        'name': 'Bob',
        'age': 21,
        'grades': {'math': 85, 'science': 90}
    }
}

# Access nested values
print("Student1 info:", students['student1'])
print("Student1 name:", students['student1']['name'])
print("Student1 math grade:", students['student1']['grades']['math'])

print("\nStudent2 name:", students['student2']['name'])
print("Student2 science grade:", students['student2']['grades']['science'])
````

### Modifying Nested Values

````python
students = {
    'student1': {
        'name': 'Alice',
        'grades': {'math': 90, 'science': 85}
    }
}

print("Before:", students)

# Modify nested value
students['student1']['grades']['math'] = 95
print("\nAfter grade change:", students)

# Add new nested key
students['student1']['grades']['english'] = 88
print("\nAfter adding english:", students)
````

### Practical Example: Company Structure

````python
company = {
    'engineering': {
        'manager': 'Alice',
        'employees': ['Bob', 'Charlie'],
        'budget': 500000
    },
    'marketing': {
        'manager': 'David',
        'employees': ['Eve', 'Frank'],
        'budget': 300000
    }
}

print("Engineering manager:", company['engineering']['manager'])
print("Marketing budget:", company['marketing']['budget'])
print("Engineering employees:", company['engineering']['employees'])
````

## 6. Dictionary from Lists

Create dictionaries from lists using `zip()` and `dict()`

### Using zip() to Combine Lists

````python
# Create dictionary from two lists
keys = ['name', 'age', 'grade']
values = ['Alice', 20, 'A']

# zip creates pairs, dict converts to dictionary
student = dict(zip(keys, values))
print("Student:", student)
````

````python
# Another example
fruits = ['apple', 'banana', 'cherry']
prices = [1.50, 0.75, 2.00]

fruit_prices = dict(zip(fruits, prices))
print("Prices:", fruit_prices)
````

### Using fromkeys() - Create with Default Value

**Syntax:** `dict.fromkeys(keys, value=None)`

````python
# Create dictionary with default values
keys = ['apple', 'banana', 'cherry']

# All values set to 0
inventory = dict.fromkeys(keys, 0)
print("Inventory:", inventory)

# All values set to empty list
student_grades = dict.fromkeys(['Alice', 'Bob', 'Charlie'], [])
print("Grades:", student_grades)
````

## 7. Reference vs Copy (IMPORTANT!)

⚠️ **CRITICAL CONCEPT** - Same as lists, this is a common source of bugs!

### The Problem: References

````python
# WRONG - This creates a reference!
dict1 = {'name': 'Alice', 'age': 20}
dict2 = dict1  # Both point to SAME dictionary

print("Before modification:")
print("dict1:", dict1)
print("dict2:", dict2)

# Modify dict2
dict2['age'] = 25

print("\nAfter modifying dict2:")
print("dict1:", dict1)  # CHANGED! Surprising!
print("dict2:", dict2)
````

### The Solution: Create a Copy

````python
# Method 1: Using .copy()
dict1 = {'name': 'Bob', 'age': 21}
dict2 = dict1.copy()

dict2['age'] = 30

print("Using .copy():")
print("dict1:", dict1)  # UNCHANGED!
print("dict2:", dict2)
````

````python
# Method 2: Using dict() constructor
dict1 = {'name': 'Charlie', 'age': 22}
dict2 = dict(dict1)

dict2['age'] = 30

print("Using dict():")
print("dict1:", dict1)  # UNCHANGED!
print("dict2:", dict2)
````

## 8. Dictionary Unpacking

Use `**` to unpack dictionaries.

### Unpacking in Function Calls

````python
# Create a dictionary
person = {'name': 'Alice', 'age': 25, 'city': 'New York'}

# Print values separately
print("Name:", person['name'])
print("Age:", person['age'])
print("City:", person['city'])
````

### Merging Dictionaries with Unpacking

````python
# Merge dictionaries using unpacking
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

# Create merged dictionary
merged = {**dict1, **dict2}
print("Merged:", merged)
````

````python
# Merge with overriding
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 20, 'c': 3}  # 'b' exists in both

merged = {**dict1, **dict2}
print("Merged:", merged)  # dict2's 'b' wins
````

### Adding Extra Items During Merge

````python
# Merge and add new items
base_info = {'name': 'Alice', 'age': 25}

full_info = {**base_info, 'email': 'alice@email.com', 'city': 'NYC'}
print("Full info:", full_info)
````

## 9. Practical Applications

Real-world uses of dictionaries.

### Counting Occurrences

````python
# Count items manually
votes = ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Alice']

# Start with empty dictionary
vote_count = {}

# Count each vote
vote_count['Alice'] = votes.count('Alice')
vote_count['Bob'] = votes.count('Bob')
vote_count['Charlie'] = votes.count('Charlie')

print("Vote counts:", vote_count)
````

### Data Lookup (Fast Access)

````python
# Store product prices
prices = {
    'apple': 1.50,
    'banana': 0.75,
    'cherry': 2.00,
    'mango': 3.50
}

# Quick lookup
print("Apple price:", prices['apple'])
print("Mango price:", prices['mango'])

# Safe lookup with default
print("Grape price:", prices.get('grape', 'Not available'))
````

### Configuration Settings

````python
# Application settings
config = {
    'theme': 'dark',
    'font_size': 14,
    'auto_save': True,
    'language': 'English'
}

print("Current theme:", config['theme'])
print("Font size:", config['font_size'])

# Update settings
config['theme'] = 'light'
config['font_size'] = 16
print("\nUpdated config:", config)
````

### Caching Results

````python
# Cache expensive calculations
cache = {}

# Calculate square
number = 5
cache[number] = number ** 2

number2 = 10
cache[number2] = number2 ** 2

print("Cache:", cache)

# Quick lookup
print("Square of 5:", cache[5])
print("Square of 10:", cache[10])
````

### Key Takeaways:

2. **Always use `.copy()`** when you want an independent copy
3. **Dictionaries are fast** - O(1) lookup time
4. **Keys must be immutable** - strings, numbers, tuples (not lists!)
5. **Nested dicts** are perfect for hierarchical data

### When to Use Dictionaries:

- **Fast lookups** by key (like phone book)
- **Counting** items (tallies, votes)
- **Grouping** related data
- **Caching** computed values
- **Configuration** settings
- **Mapping** relationships (word → definition, product → price)

### Common Patterns:

```python
# Safe access with default
value = my_dict.get(key, default)

# Check before access
if key in my_dict:
    value = my_dict[key]

# Initialize with default
my_dict.setdefault(key, default_value)

# Merge dictionaries
merged = {**dict1, **dict2}

# Create independent copy
new_dict = old_dict.copy()
```
