## 1. Additional Set Methods

Beyond `add()` and `remove()`, Python sets have many useful methods.

### discard() - Safe Removal

**Syntax:** `set.discard(item)`

Like `remove()` but doesn't raise an error if item doesn't exist

````python
# discard() won't cause errors
fruits = {'apple', 'banana', 'cherry'}
print("Original:", fruits)

fruits.discard('banana')
print("After discarding 'banana':", fruits)

# This won't cause an error (unlike remove)
fruits.discard('mango')  # Not in set, but no error!
print("After discarding 'mango':", fruits)
````

### pop() - Remove Random Element

**Syntax:** `set.pop()`

Removes and returns an arbitrary element

````python
# pop() removes an arbitrary element
numbers = {1, 2, 3, 4, 5}
print("Original:", numbers)

removed = numbers.pop()
print("Popped:", removed)
print("After pop:", numbers)
````

### clear() - Remove All Elements

**Syntax:** `set.clear()`

````python
# clear() empties the set
numbers = {1, 2, 3, 4, 5}
print("Before clear:", numbers)

numbers.clear()
print("After clear:", numbers)
````

### copy() - Create a Copy

**Syntax:** `new_set = set.copy()`

````python
# copy() creates an independent copy
original = {1, 2, 3}
copy_set = original.copy()

print("Original:", original)
print("Copy:", copy_set)

# Modifying copy doesn't affect original
copy_set.add(4)
print("\nAfter modifying copy:")
print("Original:", original)
print("Copy:", copy_set)
````

## 2. Set Operations - Methods vs Operators

Set operations can be performed using either methods or operators.

### Union - Combine Sets

**Method:** `set1.union(set2)`  
**Operator:** `set1 | set2`

````python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Using method
result1 = set1.union(set2)
print("Union (method):", result1)

# Using operator
result2 = set1 | set2
print("Union (operator):", result2)
````

### Intersection - Common Elements

**Method:** `set1.intersection(set2)`  
**Operator:** `set1 & set2`

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Using method
result1 = set1.intersection(set2)
print("Intersection (method):", result1)

# Using operator
result2 = set1 & set2
print("Intersection (operator):", result2)
````

### Difference - Elements in First but Not Second

**Method:** `set1.difference(set2)`  
**Operator:** `set1 - set2`

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Using method
result1 = set1.difference(set2)
print("Difference (method):", result1)

# Using operator
result2 = set1 - set2
print("Difference (operator):", result2)
````

### Symmetric Difference - Elements in Either but Not Both

**Method:** `set1.symmetric_difference(set2)`  
**Operator:** `set1 ^ set2`

````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Using method
result1 = set1.symmetric_difference(set2)
print("Symmetric Difference (method):", result1)

# Using operator
result2 = set1 ^ set2
print("Symmetric Difference (operator):", result2)
````

## 3. Modifying Sets with Update Methods

These methods modify the set in-place.

### update() - Add Multiple Elements

**Syntax:** `set.update(iterable)`

````python
# update() adds multiple elements
numbers = {1, 2, 3}
print("Original:", numbers)

numbers.update([4, 5, 6])
print("After update with list:", numbers)

numbers.update({7, 8, 9})
print("After update with set:", numbers)
````

### intersection_update() - Keep Only Common Elements

**Syntax:** `set1.intersection_update(set2)`

````python
# intersection_update() modifies the set in place
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}

print("Before:", set1)
set1.intersection_update(set2)
print("After intersection_update:", set1)
````

### difference_update() - Remove Common Elements

**Syntax:** `set1.difference_update(set2)`

````python
# difference_update() removes elements found in other set
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}

print("Before:", set1)
set1.difference_update(set2)
print("After difference_update:", set1)
````

### symmetric_difference_update() - Keep Only Unique Elements

**Syntax:** `set1.symmetric_difference_update(set2)`

````python
# symmetric_difference_update() keeps elements in either but not both
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}

print("Before:", set1)
set1.symmetric_difference_update(set2)
print("After symmetric_difference_update:", set1)
````

## 4. Set Comparison Methods

Check relationships between sets.

### issubset() - Check if Subset

**Syntax:** `set1.issubset(set2)`  
**Operator:** `set1 <= set2`

````python
# Check if one set is subset of another
small = {1, 2}
large = {1, 2, 3, 4, 5}

print("Is small subset of large?", small.issubset(large))
print("Using operator:", small <= large)

print("\nIs large subset of small?", large.issubset(small))
````

### issuperset() - Check if Superset

**Syntax:** `set1.issuperset(set2)`  
**Operator:** `set1 >= set2`

````python
# Check if one set is superset of another
small = {1, 2}
large = {1, 2, 3, 4, 5}

print("Is large superset of small?", large.issuperset(small))
print("Using operator:", large >= small)

print("\nIs small superset of large?", small.issuperset(large))
````

### isdisjoint() - Check if No Common Elements

**Syntax:** `set1.isdisjoint(set2)`

````python
# Check if sets have no elements in common
set1 = {1, 2, 3}
set2 = {4, 5, 6}
set3 = {3, 4, 5}

print("Are set1 and set2 disjoint?", set1.isdisjoint(set2))
print("Are set1 and set3 disjoint?", set1.isdisjoint(set3))
````

## 5. Membership Testing and Length

Check if elements exist and get set size.

### 'in' Operator - Very Fast!

Sets are optimized for membership testing (O(1) time complexity)

````python
# Check if element exists
fruits = {'apple', 'banana', 'cherry'}

print("Is 'apple' in set?", 'apple' in fruits)
print("Is 'mango' in set?", 'mango' in fruits)

# Check if NOT in set
print("Is 'grape' not in set?", 'grape' not in fruits)
````

### len() - Get Size

````python
# Get number of elements
numbers = {1, 2, 3, 4, 5}
print("Set:", numbers)
print("Length:", len(numbers))

# Empty set
empty = set()
print("\nEmpty set length:", len(empty))
````

## 6. Converting to and from Sets

Sets can be created from and converted to other data types.

### List to Set - Remove Duplicates!

````python
# Convert list to set (removes duplicates)
numbers_list = [1, 2, 2, 3, 3, 3, 4, 4, 5]
print("Original list:", numbers_list)

numbers_set = set(numbers_list)
print("As set:", numbers_set)
print("Duplicates removed!")
````

### Set to List - Get Ordered Version

````python
# Convert set to list
my_set = {3, 1, 4, 1, 5, 9, 2, 6}
print("Set:", my_set)

my_list = list(my_set)
print("As list:", my_list)

# Sort the list
my_list.sort()
print("Sorted list:", my_list)
````

### String to Set - Unique Characters

````python
# Convert string to set (gets unique characters)
text = "hello world"
print("String:", text)

unique_chars = set(text)
print("Unique characters:", unique_chars)
print("Number of unique chars:", len(unique_chars))
````

### Tuple to Set

````python
# Convert tuple to set
my_tuple = (1, 2, 2, 3, 3, 3)
print("Tuple:", my_tuple)

my_set = set(my_tuple)
print("As set:", my_set)
````

## 7. Practical Applications

Real-world uses of sets.

### Removing Duplicates from Data

````python
# Remove duplicate emails
email_list = [
    'alice@email.com',
    'bob@email.com',
    'alice@email.com',  # duplicate
    'charlie@email.com',
    'bob@email.com'     # duplicate
]

print("Original emails:", email_list)
print("Count:", len(email_list))

unique_emails = list(set(email_list))
print("\nUnique emails:", unique_emails)
print("Count:", len(unique_emails))
````

### Finding Common Elements

````python
# Find common interests between friends
alice_interests = {'music', 'sports', 'reading', 'gaming'}
bob_interests = {'movies', 'sports', 'gaming', 'cooking'}

common_interests = alice_interests & bob_interests
print("Common interests:", common_interests)
````

### Finding Unique Elements

````python
# Find unique skills needed
job1_skills = {'Python', 'SQL', 'Excel', 'Communication'}
job2_skills = {'Python', 'JavaScript', 'Git', 'Communication'}

# Skills unique to each job
job1_only = job1_skills - job2_skills
job2_only = job2_skills - job1_skills

print("Job 1 unique skills:", job1_only)
print("Job 2 unique skills:", job2_only)

# All skills needed for either job
all_skills = job1_skills | job2_skills
print("All skills:", all_skills)
````

### Checking Data Quality

````python
# Check for duplicate IDs
user_ids = [101, 102, 103, 104, 102, 105, 103]  # Has duplicates

unique_ids = set(user_ids)

print("Total IDs:", len(user_ids))
print("Unique IDs:", len(unique_ids))

has_duplicates = len(user_ids) != len(unique_ids)
print("Has duplicates?", has_duplicates)
````

## 8. Set Limitations

Understanding what sets cannot do.

### No Indexing

````python
# Sets are unordered - no indexing!
my_set = {1, 2, 3, 4, 5}

# This will cause an error:
# print(my_set[0])  # TypeError!

# Convert to list if you need indexing
my_list = list(my_set)
print("First element:", my_list[0])
````

### No Duplicate Values

````python
# Sets automatically remove duplicates
my_set = {1, 2, 2, 3, 3, 3}
print("Set:", my_set)  # Only unique values remain
````

### No Mutable Elements

Sets cannot contain lists or dictionaries (only immutable types)

````python
# These work (immutable types):
valid_set = {1, 'hello', 3.14, (1, 2), True}
print("Valid set:", valid_set)

# This will cause an error:
# invalid_set = {1, 2, [3, 4]}  # TypeError! Lists are mutable
# invalid_set = {1, 2, {'key': 'value'}}  # TypeError! Dicts are mutable
````

### When to Use Sets:

- Remove duplicate values from data
- Fast membership testing ("is X in the collection?")

- Finding unique elements between groups
- Mathematical set operations

### Set Operations Quick Reference:

| Operation | Method | Operator | Description |
|-----------|--------|----------|-------------|

| Difference | `difference()` | `-` | Elements in first but not second |
