### 1.4 Keyword Arguments

Call functions using parameter names - order doesn't matter!

**Benefit:** Makes code more readable and flexible.

````python
# Example 1: Positional vs Keyword arguments
def describe_person(name, age, city):
    print(f"{name} is {age} years old and lives in {city}")

# Positional arguments (order matters)
describe_person("Alice", 25, "New York")

# Keyword arguments (order doesn't matter)
describe_person(age=30, city="London", name="Bob")
describe_person(name="Charlie", age=35, city="Paris")
````

````python
# Example 2: Mix positional and keyword
def create_account(username, password, email="", phone=""):
    print(f"Username: {username}")
    print(f"Password: {password}")
    print(f"Email: {email}")
    print(f"Phone: {phone}")
    print()

# Positional for required, keyword for optional
create_account("alice123", "pass123")
create_account("bob456", "secret", email="bob@email.com")
create_account("charlie", "pass", phone="123-456", email="c@mail.com")
````

````python
# Example 3: Makes code more readable
def draw_rectangle(width, height, color="black", filled=False):
    print(f"Drawing rectangle:")
    print(f"  Size: {width}x{height}")
    print(f"  Color: {color}")
    print(f"  Filled: {filled}")
    print()

# Clear and readable
draw_rectangle(width=100, height=50, color="red", filled=True)
draw_rectangle(height=75, width=150)  # Can change order!
````

## Section 2: Return Statements

Functions can send back values using `return`.

### 2.1 Basic Return

````python
# Example 1: Return a value
def add(a, b):
    return a + b

result = add(5, 3)
print(f"Result: {result}")
````

````python
# Example 2: Use return value in expression
def square(n):
    return n ** 2

print(f"5 squared: {square(5)}")
print(f"Sum of squares: {square(3) + square(4)}")
````

````python
# Example 3: Return with condition
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

print(f"4 is even: {is_even(4)}")
print(f"7 is even: {is_even(7)}")
````

### 2.2 None as Return Value

Functions without explicit return (or `return` without value) return `None`.

````python
# Example 1: No return statement
def print_message(msg):
    print(msg)

result = print_message("Hello")
print(f"Return value: {result}")
print(f"Type: {type(result)}")
````

````python
# Example 2: Explicit return None
def find_item(items, target):
    for item in items:
        if item == target:
            return item
    return None  # Explicit: not found

result1 = find_item([1, 2, 3], 2)
result2 = find_item([1, 2, 3], 5)

print(f"Found: {result1}")
print(f"Not found: {result2}")
````

````python
# Example 3: Check if None
def divide(a, b):
    if b == 0:
        return None
    return a / b

result = divide(10, 2)
if result is not None:
    print(f"Result: {result}")

result = divide(10, 0)
if result is None:
    print("Cannot divide by zero")
````

### 2.3 Returning Multiple Values

Functions can return multiple values as a tuple.

**Syntax:** `return value1, value2, value3`

````python
# Example 1: Return two values
def get_name_age():
    return "Alice", 25

# Tuple unpacking
name, age = get_name_age()
print(f"Name: {name}, Age: {age}")
````

````python
# Example 2: Return three values
def get_min_max_avg(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)
    return minimum, maximum, average

data = [10, 20, 30, 40, 50]
min_val, max_val, avg_val = get_min_max_avg(data)

print(f"Min: {min_val}")
print(f"Max: {max_val}")
print(f"Average: {avg_val}")
````

````python
# Example 3: Return as tuple (explicit)
def divide_with_remainder(a, b):
    quotient = a // b
    remainder = a % b
    return quotient, remainder

q, r = divide_with_remainder(17, 5)
print(f"17 ÷ 5 = {q} remainder {r}")
````

````python
# Example 4: Can also capture as tuple
def get_coordinates():
    return 10, 20, 30

# As tuple
coords = get_coordinates()
print(f"Coordinates: {coords}")
print(f"Type: {type(coords)}")

# Or unpack
x, y, z = get_coordinates()
print(f"x={x}, y={y}, z={z}")
````

## Section 3: Variable Scope

Where variables can be accessed: local vs global scope.

### 3.1 Local Scope

````python
# Example 1: Local variable
def my_function():
    x = 10  # Local to this function
    print(f"Inside function: x = {x}")

my_function()
# print(x)  # Error! x doesn't exist outside function
````

### 3.2 Global Scope

````python
# Example 2: Global variable
x = 100  # Global variable

def my_function():
    print(f"Inside function: x = {x}")  # Can read global

my_function()
print(f"Outside function: x = {x}")
````

### 3.3 The global Keyword

Modify global variables inside functions using `global` keyword.

````python
# Example 3: Without global (creates local variable)
counter = 0

def increment():
    counter = counter + 1  # Error! References before assignment
    print(counter)

# increment()  # This would cause an error
````

````python
# Example 4: With global keyword
counter = 0

def increment():
    global counter
    counter = counter + 1
    print(f"Counter: {counter}")

increment()
increment()
increment()
print(f"Final counter: {counter}")
````

````python
# Example 5: Multiple global variables
total = 0
count = 0

def add_score(score):
    global total, count
    total += score
    count += 1
    average = total / count
    print(f"Average: {average}")

add_score(85)
add_score(90)
add_score(78)
````

## Section 4: Docstrings

Document your functions with docstrings - helps others (and future you!) understand what the function does.

**Syntax:** Triple quotes `"""` immediately after function definition.

````python
# Example 1: Simple docstring
def greet(name):
    """Prints a greeting message with the given name."""
    print(f"Hello, {name}!")

# Access docstring
print(greet.__doc__)
help(greet)
````

````python
# Example 2: Detailed docstring
def calculate_area(length, width):
    """
    Calculate the area of a rectangle.
    
    Parameters:
    length (float): The length of the rectangle
    width (float): The width of the rectangle
    
    Returns:
    float: The area of the rectangle
    """
    return length * width

print(calculate_area.__doc__)
result = calculate_area(5, 10)
print(f"Area: {result}")
````

````python
# Example 3: Comprehensive docstring
def find_max(numbers):
    """
    Find the maximum value in a list of numbers.
    
    Parameters:
    numbers (list): A list of numeric values
    
    Returns:
    float/int: The maximum value in the list
    None: If the list is empty
    
    Example:
    >>> find_max([1, 5, 3, 9, 2])
    9
    """
    if not numbers:
        return None
    return max(numbers)

help(find_max)
````

## Section 5: Variable Arguments

Functions that accept any number of arguments.

### 5.1 *args - Variable Positional Arguments

Accept any number of positional arguments as a tuple.

**Syntax:** `def function(*args):`

````python
# Example 1: Sum any number of values
def add_all(*numbers):
    """Add all provided numbers."""
    total = 0
    for num in numbers:
        total += num
    return total

print(add_all(1, 2, 3))           # 3 arguments
print(add_all(10, 20, 30, 40))    # 4 arguments
print(add_all(5))                  # 1 argument
````

````python
# Example 2: Understanding *args
def print_args(*args):
    print(f"Type: {type(args)}")
    print(f"Values: {args}")
    for i, arg in enumerate(args):
        print(f"  Argument {i}: {arg}")

print_args("apple", "banana", "cherry")
````

````python
# Example 3: Mix regular and *args
def make_sentence(start, *words):
    """Create a sentence starting with 'start' and adding all words."""
    sentence = start
    for word in words:
        sentence += " " + word
    return sentence

print(make_sentence("Hello", "world", "how", "are", "you"))
print(make_sentence("Python", "is", "awesome"))
````

````python
# Example 4: Find maximum of any count of numbers
def find_max(*numbers):
    """Find maximum among any number of values."""
    if not numbers:
        return None
    return max(numbers)

print(find_max(5, 2, 9, 1, 7))
print(find_max(10, 25))
print(find_max(42))
````

### 5.2 **kwargs - Variable Keyword Arguments

Accept any number of keyword arguments as a dictionary.

**Syntax:** `def function(**kwargs):`

````python
# Example 1: Basic **kwargs
def print_info(**kwargs):
    """Print all keyword arguments."""
    print(f"Type: {type(kwargs)}")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=25, city="New York")
print()
print_info(product="Laptop", price=999, brand="TechCo")
````

````python
# Example 2: Create profile with any attributes
def create_profile(**details):
    """Create user profile with any attributes."""
    print("User Profile:")
    for key, value in details.items():
        print(f"  {key.title()}: {value}")
    print()

create_profile(name="Alice", age=30, email="alice@email.com")
create_profile(username="bob123", city="London", country="UK", status="active")
````

````python
# Example 3: Mix regular parameters and **kwargs
def configure_settings(theme, **options):
    """Configure app with theme and any additional options."""
    print(f"Theme: {theme}")
    print("Additional settings:")
    for key, value in options.items():
        print(f"  {key}: {value}")

configure_settings("dark", language="English", font_size=14, notifications=True)
````

````python
# Example 4: Combining *args and **kwargs
def super_function(*args, **kwargs):
    """Accepts any positional and keyword arguments."""
    print("Positional arguments:")
    for arg in args:
        print(f"  {arg}")
    
    print("\nKeyword arguments:")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

super_function(1, 2, 3, name="Alice", age=25)
````

## Section 6: Functions as Objects

In Python, functions are first-class objects - they can be assigned to variables, passed as arguments, and returned from other functions.

### 6.1 Assigning Functions to Variables

````python
# Example 1: Store function in variable
def greet(name):
    return f"Hello, {name}!"

# Assign to variable (no parentheses!)
say_hello = greet

# Call through variable
print(say_hello("Alice"))
print(greet("Bob"))
````

### 6.2 Passing Functions as Arguments

````python
# Example 2: Function as parameter
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def calculate(operation, x, y):
    """Perform operation on x and y."""
    result = operation(x, y)
    return result

# Pass functions as arguments
print(calculate(add, 5, 3))        # 8
print(calculate(multiply, 5, 3))   # 15
````

````python
# Example 3: Apply function to list
def square(n):
    return n ** 2

def cube(n):
    return n ** 3

def apply_to_list(func, numbers):
    """Apply function to each number in list."""
    result = []
    for num in numbers:
        result.append(func(num))
    return result

data = [1, 2, 3, 4, 5]
print("Squares:", apply_to_list(square, data))
print("Cubes:", apply_to_list(cube, data))
````

````python
# Example 4: Flexible filtering
def is_even(n):
    return n % 2 == 0

def is_positive(n):
    return n > 0

def filter_list(condition, numbers):
    """Filter list based on condition function."""
    result = []
    for num in numbers:
        if condition(num):
            result.append(num)
    return result

data = [-2, -1, 0, 1, 2, 3, 4, 5]
print("Even numbers:", filter_list(is_even, data))
print("Positive numbers:", filter_list(is_positive, data))
````

## Section 7: Lambda Functions

Small anonymous functions defined in one line.

**Syntax:** `lambda arguments: expression`

````python
# Example 1: Basic lambda
square = lambda x: x ** 2

print(square(5))
print(square(10))
````

````python
# Example 2: Lambda with multiple arguments
add = lambda a, b: a + b
multiply = lambda a, b: a * b

print(add(3, 5))
print(multiply(3, 5))
````

````python
# Example 3: Lambda as argument
def apply_operation(func, x, y):
    return func(x, y)

result = apply_operation(lambda a, b: a + b, 10, 20)
print(result)
````

````python
# Example 4: Lambda with sorting
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 92},
    {'name': 'Charlie', 'grade': 78}
]

# Sort by grade using lambda
sorted_students = sorted(students, key=lambda s: s['grade'])

for student in sorted_students:
    print(f"{student['name']}: {student['grade']}")
````
