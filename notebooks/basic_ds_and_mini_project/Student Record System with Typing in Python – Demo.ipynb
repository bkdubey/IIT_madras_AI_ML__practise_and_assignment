```
from typing import Dict, Tuple, List, Set
```

```
StudentID = int
StudentRecord = Tuple[str, Dict[str, int]]  #(name, {subject : marks})
Database = Dict[StudentID, StudentRecord]
```

```
students_db : Database = {}
```

```
def add_student(db:Database) -> None:
  student_id : int = int(input("Enter student id: "))
  if student_id in db:
    print("Student already exists")
    return

  name : str = input("Enter student name: ")
  subject_count : int = int(input("Enter number of subjects: "))

  subjects : Dict[str, int] = {}
  for _ in range(subject_count):
    subject : str = input("Enter subject name: ")
    marks : int = int(input(f"Enter marks for {subject}: "))
    subjects[subject] = marks

  db[student_id] = (name, subjects)
  print("Student added successfully")
```

```
def view_student(db:Database) -> None:
  if not db:
    print("No students in database")
    return

  for student_id, (name, marks) in db.items():
    print(f"/n ID : {student_id}, Name : {name}")
    for subject, marks in marks.items():
      print(f"Subject ({subject}) : Marks ({marks})")
```

```
def search_student(db:Database) -> None:
  student_id : int = int(input("Enter student id: "))
  if student_id not in db:
    print("Student not found")
    return

  name, marks = db[student_id]
  print(f"/n ID : {student_id}, Name : {name}")
  for subject, marks in marks.items():
    print(f"Subject ({subject}) : Marks ({marks})")
```

```
def run_app() -> None:
  while True:
    print("1. Add student")
    print("2. View student")
    print("3. Search student")
    print("4. Exit")

    choice : int = int(input("Enter your choice: "))

    if choice == 1:
      add_student(students_db)
    elif choice == 2:
      view_student(students_db)
    elif choice == 3:
      search_student(students_db)
    elif choice == 4:
      break
```

```
run_app()
```

```
def delete_student(db: Database) -> None:
    student_id = int(input("Enter ID to delete: "))
    if student_id in db:
        del db[student_id]
        print(" Deleted successfully.")
    else:
        print(" ID not found.")

def get_all_subjects(db: Database) -> Set[str]:
    all_subjects: Set[str] = set()
    for _, (_, marks) in db.items():
        all_subjects.update(marks.keys())
    return all_subjects

def view_subjects(db: Database) -> None:
    subjects = get_all_subjects(db)
    if not subjects:
        print("No subjects found.")
    else:
        print(" Unique Subjects:")
        for subject in sorted(subjects):
            print(" -", subject)

# Main loop
def run_app() -> None:
    while True:
        print("\n--- Student Record System ---")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search by ID")
        print("4. Delete Student")
        print("5. View Unique Subjects")
        print("6. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            add_student(students_db)
        elif choice == "2":
            view_all_students(students_db)
        elif choice == "3":
            search_student(students_db)
        elif choice == "4":
            delete_student(students_db)
        elif choice == "5":
            view_subjects(students_db)
        elif choice == "6":
            print(" Exiting the system. Goodbye!")
            break
        else:
            print("❗ Invalid choice. Try again.")

# Run the application
run_app()
```

```
from typing import Dict, Tuple, List, Set, Optional

StudentID = int
StudentRecord = Tuple[str, Dict[str, int]]  # (name, {subject: mark})
Database = Dict[StudentID, StudentRecord]
```

```
def calculate_grade(average: float) -> str:
    if average >= 90: return 'A+'
    elif average >= 80: return 'A'
    elif average >= 70: return 'B'
    elif average >= 60: return 'C'
    else: return 'D'
```

```
def view_all_students(db: Database) -> None:
    if not db:
        print("No student records found.")
        return

    for student_id, (name, marks) in db.items():
        print(f"\n ID: {student_id}, Name: {name}")
        total = sum(marks.values())
        avg = total / len(marks)
        grade = calculate_grade(avg)
        for subject, mark in marks.items():
            print(f"   {subject}: {mark} marks")
        print(f"   Average: {avg:.2f}, Grade: {grade}")
```

```
def view_topper(db: Database) -> None:
    if not db:
        print("No students to evaluate.")
        return

    def avg_marks(record: StudentRecord) -> float:
        return sum(record[1].values()) / len(record[1])

    topper_id = max(db, key=lambda sid: avg_marks(db[sid]))
    name, marks = db[topper_id]
    avg = avg_marks((name, marks))
    print(f"\nTopper is {name} (ID: {topper_id}) with Average Marks: {avg:.2f}")
```

```
def average_per_subject(db: Database) -> None:
    subject_totals: Dict[str, List[int]] = {}

    for _, (_, marks) in db.items():
        for subject, score in marks.items():
            subject_totals.setdefault(subject, []).append(score)

    print("Subject-wise Averages:")
    for subject, scores in subject_totals.items():
        avg = sum(scores) / len(scores)
        print(f" - {subject}: {avg:.2f}")
```

```
def run_app() -> None:
    while True:
        print("\n--- Student Record System ---")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search by ID")
        print("4. Delete Student")
        print("5. View Unique Subjects")
        print("6. View Topper")
        print("7. Subject-wise Average")
        print("8. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            add_student(students_db)
        elif choice == "2":
            view_all_students(students_db)
        elif choice == "3":
            search_student(students_db)
        elif choice == "4":
            delete_student(students_db)
        elif choice == "5":
            view_subjects(students_db)
        elif choice == "6":
            view_topper(students_db)
        elif choice == "7":
            average_per_subject(students_db)
        elif choice == "8":
            print("Exiting. Goodbye!")
            break
        else:
            print("Invalid choice. Try again.")
```
