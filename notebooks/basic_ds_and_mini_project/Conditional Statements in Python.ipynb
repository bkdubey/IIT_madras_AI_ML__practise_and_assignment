## Section 4: Comparison Operators

## Section 5: Logical Operators

Combine multiple conditions using logical operators:

| Operator | Description | Example |
|----------|-------------|--------|
| `and` | Both conditions must be True | `x > 0 and x < 10` |
| `or` | At least one condition must be True | `x < 0 or x > 10` |
| `not` | Reverses the condition | `not x == 5` |

### 5.1 The AND Operator

Both conditions must be True for the entire expression to be True.

````python
# Example 1: Age and citizenship check
age = 25
is_citizen = True

if age >= 18 and is_citizen:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
````

````python
# Example 2: Range check
score = 85

if score >= 80 and score <= 90:
    print("Grade: B")
````

````python
# Example 3: Multiple conditions
temperature = 22
humidity = 60
is_raining = False

if temperature > 20 and humidity < 70 and not is_raining:
    print("Perfect weather for a picnic!")
````

### 5.2 The OR Operator

At least one condition must be True for the entire expression to be True.

````python
# Example 1: Weekend check
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a weekday")
````

````python
# Example 2: Discount eligibility
age = 65
is_student = False

if age < 18 or age >= 65 or is_student:
    print("You get a discount!")
else:
    print("Regular price")
````

### 5.3 The NOT Operator

Reverses the boolean value - True becomes False, False becomes True.

````python
# Example 1: NOT operator
is_raining = False

if not is_raining:
    print("Go for a walk!")
````

````python
# Example 2: NOT with comparison
password = "abc123"

if not password == "correct":
    print("Incorrect password")
````

### 5.4 Combining Logical Operators

````python
# Example: Complex condition
age = 25
has_license = True
has_car = True
is_insured = True

if age >= 18 and has_license and (has_car or is_insured):
    print("Can drive legally")
else:
    print("Cannot drive")
````

## Section 6: Membership Operators

Check if a value exists in a sequence (list, tuple, string, etc.).

| Operator | Description |
|----------|-------------|
| `in` | Returns True if value exists in sequence |
| `not in` | Returns True if value doesn't exist in sequence |

````python
# Example 1: Check if item in list
fruits = ['apple', 'banana', 'cherry']

if 'apple' in fruits:
    print("We have apples!")

if 'mango' not in fruits:
    print("No mangoes available")
````

````python
# Example 2: Check character in string
email = "user@example.com"

if '@' in email and '.' in email:
    print("Valid email format")
else:
    print("Invalid email")
````

````python
# Example 3: Check key in dictionary
student = {'name': 'Alice', 'age': 20, 'grade': 'A'}

if 'name' in student:
    print(f"Student name: {student['name']}")

if 'email' not in student:
    print("Email not provided")
````

## Section 7: Identity Operators

Compare object identity (whether they are the same object in memory).

| Operator | Description |
|----------|-------------|
| `is` | Returns True if both refer to the same object |
| `is not` | Returns True if they refer to different objects |

### 7.1 Understanding `is` vs `==`

- `==` checks if values are equal
- `is` checks if they are the same object in memory

````python
# Example 1: is vs ==
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)  # True (same values)
print("a is b:", a is b)  # False (different objects)
print("a is c:", a is c)  # True (same object)
````

````python
# Example 2: Common use with None
value = None

if value is None:
    print("No value provided")

result = "something"
if result is not None:
    print("Result exists")
````

````python
# Example 3: Boolean singletons
flag = True

if flag is True:  # Usually just: if flag:
    print("Flag is true")

status = False
if status is False:  # Usually just: if not status:
    print("Status is false")
````

## Section 8: Truthy and Falsy Values

In Python, values can be evaluated as True or False in a boolean context.

**Falsy values** (evaluate to False):
- `False`
- `None`
- `0` (zero)
- `0.0` (zero float)
- `''` (empty string)
- `[]` (empty list)
- `()` (empty tuple)
- `{}` (empty dictionary)
- `set()` (empty set)

**Truthy values** (evaluate to True):
- Everything else!

````python
# Example 1: Checking empty sequences
my_list = []

if my_list:
    print("List has items")
else:
    print("List is empty")
````

````python
# Example 2: Checking strings
username = ""

if username:
    print(f"Welcome, {username}")
else:
    print("Please enter a username")
````

````python
# Example 3: Checking numbers
count = 0

if count:
    print(f"You have {count} items")
else:
    print("No items")
````

````python
# Example 4: Checking None
result = None

if result:
    print(f"Result: {result}")
else:
    print("No result yet")
````

````python
# Example 5: Truthy values
name = "Alice"
items = [1, 2, 3]
count = 5

if name and items and count:
    print("All values are truthy")
````

## Section 10: Ternary Operator (Conditional Expression)

A compact way to write simple if-else statements in one line.

**Syntax:**
```python
value_if_true if condition else value_if_false
```

````python
# Example 1: Traditional if-else
age = 20

if age >= 18:
    status = "Adult"
else:
    status = "Minor"

print(status)
````

````python
# Example 2: Same with ternary operator
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
````

````python
# Example 3: Finding maximum
a = 10
b = 20

max_value = a if a > b else b
print(f"Maximum: {max_value}")
````

````python
# Example 4: Even or odd
number = 7
result = "Even" if number % 2 == 0 else "Odd"
print(f"{number} is {result}")
````

````python
# Example 5: In print statement
score = 75
print("Passed" if score >= 60 else "Failed")
````

## Section 11: Nested Conditionals

If statements inside other if statements for complex decision-making.

````python
# Example 1: Age and membership check
age = 25
has_membership = True

if age >= 18:
    if has_membership:
        print("Access granted with member benefits")
    else:
        print("Access granted - Regular rate")
else:
    print("Access denied - Under 18")
````

````python
# Example 2: Grade classification with attendance
score = 85
attendance = 90

if score >= 60:
    if attendance >= 75:
        print("Pass")
    else:
        print("Fail - Low attendance")
else:
    print("Fail - Low score")
````

````python
# Example 3: Can often be simplified with 'and'
score = 85
attendance = 90

if score >= 60 and attendance >= 75:
    print("Pass")
else:
    print("Fail")
````

### What You Practiced:

✅ **Basic conditionals**: if, elif, else  
✅ **Comparison operators**: ==, !=, <, >, <=, >=  
✅ **Logical operators**: and, or, not  
✅ **Membership operators**: in, not in  
✅ **Identity operators**: is, is not

### Key Takeaways:

1. **Use `and` when ALL conditions must be True**
2. **Use `or` when ANY condition can be True**
3. **`is` checks identity, `==` checks equality**
4. **Empty sequences and 0 are falsy, everything else is truthy**
5. **Ternary operator makes simple if-else more concise**
6. **Chained comparisons are cleaner: `10 <= x <= 20`**
7. **Use `in` to check membership efficiently**

### Best Practices:

- Keep conditions simple and readable
- Use logical operators instead of deeply nested ifs
- Check for None using `is None`, not `== None`
- Use truthy/falsy for existence checks
- Use ternary operator only for simple cases

**Keep practicing!** Conditional logic is fundamental to all programming.
